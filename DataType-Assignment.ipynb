	Input:
	list1 = [1,2,3,4,5,6]


	Output:
	[1,3,6,10,15,21]


```python

list1 = [1,2,3,4,5,6]

sum_list = []
total = 0

for i in list1:
    total += i
    sum_list.append(total)
print(sum_list)


```

	Input:
	Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"


	Output:
	No of unique vowels-6


```python

Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"

vowels = set('aeiouAEIOU')

unique_vowels = set([i for i in Str1 if i in vowels])

print('No of unique vowels -', len(unique_vowels))


```

17. Write a Python function to concatenate any no of dictionaries to create a new one.
	Sample Dictionary :
	dic1={1:10, 2:20}
	dic2={3:30, 4:40}
	dic3={5:50, 6:60}
	Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}


```python

dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50, 6:60}

def con_dict(*dicts):
    result = {}
    for dict in dicts:
        result.update(dict)
    return result

print(con_dict(dic1,dic2,dic3))



# 18. Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
